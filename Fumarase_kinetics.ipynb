# Enzyme Kinetics Analysis: Fumarase

This notebook is dedicated to the comprehensive analysis of enzyme kinetics, specifically focusing on the enzyme **fumarase**. We will first determine enzyme activity for each of your malate concentrations in Units/mg. Following this, we will delve into the fundamental principles of Michaelis-Menten kinetics, utilizing experimental data to determine key kinetic parameters such as Km and Vmax through non-linear curve fitting. Finally, we will employ Lineweaver-Burk and Eadie-Hofstee plots and linear regression and will determine Km and Vmax from these plots.

## Calculation of Activity in U/mg

To determine your enzyme actvity, you will enter your average velocity vs malate concentration data in units of deltaAU/min. These results also should have been corrected to a 1 cm pathlength. You should have calculated these results last week.

The data should be entered as a Python list of tuples, where each tuple represents a well and contains the following:

`(Row Identifier on your plate, Activity in deltaAU/min, concentration of malate in mM)`.

We have prefilled the Rows and malate concentrations. If you made changes to your plate, please adjust the code below. After entering your data, run the code cell below by clicking the 'play button'. The subsequent code will process your input to determine the activity in umol/min.

```python
# Define a list of tuples named fumarate_abs_data to store the experimental data for fumarate.
# Each tuple should represent a well and contain:(Row Identifier on your plate, Activity in deltaAU/min, concentration of malate in mM).
# XXX data reflects where you should enter your average activity data in deltaAU/min (no units, just the number).
fumarase_abs_data = [("A", XXX, 0.0), # Control - with no enzyme added
                    ("B", XXX, 0.05), # 0.05 mM malate
                    ("C", XXX, 0.1), # 0.1 mM malate
                    ("D", XXX, 0.25), # 0.25 mM malate
                    ("E", XXX, 0.5), # 0.5 mM malate
                    ("F", XXX, 1.0), # 1.0 mM malate
                    ("G", XXX, 2.0)] # 2.0 mM malate


# Initialize the molextinction variable to 1.440 (mM-1 cm-1),
# which is the molar extinction coefficient for fumarate at 250 nm.
molextinction = 1.440 # in mM-1 cm-1

# set the volume of the reaction in mL
volume = 0.2 # mL

# Initialize an empty list called activity to store the calculated activities.
activity = []

# Iterate through the fumarate_abs_data list to calculate enzyme activity.
for item in fumarase_abs_data:
  # item[0] is Row Identifier, item[1] is deltaAU/min, item[2] is [malate] in mM
  act = item[1] * volume/(molextinction)
  rounded = round(act, 4)

  # Store the row identifier and the rounded activity as a tuple in the activity list.
  data1 = (item[0], rounded, item[2])
  activity.append(data1)

  # Print the row identifier and the calculated activity in a user-friendly format.
  print(f"Row: {item[0]} \t Activity: {rounded} umol/min")
```

## Convert each activity to U/mg

Since we now have our data in umol/min, which equals a Unit of enzyme activity (U), we now simply need to divide each of our activity data by the mg of enzyme added. You should be able to do this calculation using your protein concentration data. Recall that you may have added 2-20 uL of enzyme and you may have had a dilution to deal with.

```python
mg_of_enzyme = float(input("Enter the mg of enzyme used in each reaction: "))

activity_per_mg = [] # List to store activities in U/mg

print("\n--- Activity per mg of enzyme ---")
for row_id, act_umol_per_min, malate in activity:
  activity_umg = act_umol_per_min / mg_of_enzyme
  rounded_activity_umg = round(activity_umg, 4)
  activity_per_mg.append((row_id, rounded_activity_umg))
  print(f"Row: {row_id}\t Activity: {rounded_activity_umg} U/mg \t [malate] {malate} mM")
print("---------------------------------")
```

## Michaelis-Menten Kinetics: Determination of Km and Vmax for Fumarase

In this section, we will delve into the quantitative analysis of fumarase's enzymatic activity by determining its key kinetic parameters, Km (Michaelis constant) and Vmax (maximum reaction velocity). This experiment involves measuring the initial reaction rates at various substrate (fumarate) concentrations.

1.  **Visualization (Michaelis-Menten Plot):** A scatter plot will be generated to visualize the relationship between substrate concentration and reaction velocity, known as a Michaelis-Menten plot. This plot will allow for a preliminary estimation of Vmax (the plateau of the curve) and Km (the substrate concentration at half Vmax).
2.  **Curve Fitting:** To obtain accurate and robust values for Km and Vmax, we will use non-linear regression (curve fitting) to fit the Michaelis-Menten equation to your experimental data. This mathematical approach provides the most precise determination of these critical kinetic parameters for fumarate.

```python
import matplotlib.pyplot as plt

# Create a dictionary to easily map row_id to malate concentration
malate_concentration_map = {item[0]: item[2] for item in activity}

# Prepare data for plotting (malate concentration, activity U/mg)
plot_data = []
for row_id, activity_umg in activity_per_mg:
    malate_conc = malate_concentration_map.get(row_id)
    if malate_conc is not None:
        plot_data.append((malate_conc, activity_umg))

# Extract x and y coordinates
x_coords = [item[0] for item in plot_data]
y_coords = [item[1] for item in plot_data]

# Create the scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(x_coords, y_coords, color='purple', label='Activity vs. Malate Concentration')

# Set labels and title
plt.xlabel('Malate Concentration (mM)')
plt.ylabel('Activity (U/mg)')
plt.title('Enzyme Activity in U/mg vs. Malate Concentration')

# Add grid and legend for better readability
plt.grid(True)
plt.legend()

# Display the plot
plt.show()
```

## Accurate Determination of Km and Vmax for Fumarase using Curve Fitting

While visual estimation provides a good initial understanding, mathematical curve fitting offers a more accurate and robust determination of the kinetic parameters Km and Vmax. We will use the Michaelis-Menten equation and non-linear regression to fit the experimental data.

Before running the next code cell, consider your visual estimates of Vmax and Km from the plot above. These values will serve as initial guesses for the curve fitting algorithm, which helps it converge to the best-fit parameters more efficiently.


```python
import numpy as np
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

# Define the Michaelis-Menten equation
def michaelis_menten(S, Vmax, Km):
  return (Vmax * S) / (Km + S)

# Perform non-linear regression using curve_fit
# Initial guesses for Vmax and Km are taken from the visual estimates
initial_guesses = [vmax_visual_estimate, km_visual_estimate]

# Filter out the 0.0 concentration point if it's causing issues for fitting
# Curve_fit will likely have issues if x_coords contains 0 and y_coords contains 0 for Michaelis-Menten
# Let's make sure we're not including the control (0 mM malate, 0 U/mg activity) in the fitting

filtered_x_coords = [x for i, x in enumerate(x_coords) if x != 0.0 or y_coords[i] != 0.0]
filtered_y_coords = [y for i, y in enumerate(y_coords) if x_coords[i] != 0.0 or y != 0.0]

params, covariance = curve_fit(michaelis_menten, filtered_x_coords, filtered_y_coords, p0=initial_guesses)

# Extract the fitted parameters
Vmax_fitted, Km_fitted = params

print(f"\n--- Fitted Michaelis-Menten Parameters ---")
print(f"Fitted Vmax: {Vmax_fitted:.4f} U/mg")
print(f"Fitted Km: {Km_fitted:.4f} mM")
print(f"----------------------------------------")

# Generate points for the fitted curve
S_fit = np.linspace(min(x_coords), max(x_coords), 100)
V_fit = michaelis_menten(S_fit, Vmax_fitted, Km_fitted)

# Plot the experimental data and the fitted curve
plt.figure(figsize=(10, 6))
plt.scatter(x_coords, y_coords, color='purple', label='Experimental Data')
plt.plot(S_fit, V_fit, color='red', linestyle='--', label='Fitted Michaelis-Menten Curve')

plt.xlabel('Malate Concentration (mM)')
plt.ylabel('Activity (U/mg)')
plt.title('Michaelis-Menten Plot: Experimental Data and Fitted Curve')
plt.grid(True)
plt.legend()
plt.show()
```

## Lineweaver-Burk Plots Allow a Linear Fit

Lineweaver-Burk plot is a double reciprocal plot of the Michaelis-Menten equation, linearizing the data.

**The Lineweaver-Burke equation is:**

$1/V = (K_m/V_{max}) * (1/[S]) + 1/V_{max}$

Where:
*   $1/V$ is the reciprocal of the reaction velocity (Y-axis)
*   $1/[S]$ is the reciprocal of the substrate concentration (X-axis)
*   $K_m/V_{max}$ is the slope of the line
*   $1/V_{max}$ is the Y-intercept
*   $-1/K_m$ is the X-intercept

By plotting $1/V$ against $1/[S]$  we can easily determine $K_m$ and $V_{max}$



```python
import numpy as np
from scipy.stats import linregress
import matplotlib.pyplot as plt

# Prepare data for Lineweaver-Burk plot
# Filter out data points where activity is 0 to avoid division by zero
filtered_lb_x = [x for i, x in enumerate(x_coords) if y_coords[i] != 0.0]
filtered_lb_y = [y for i, y in enumerate(y_coords) if y != 0.0]

# Calculate reciprocals
reciprocal_S = np.array([1/s for s in filtered_lb_x])
reciprocal_V = np.array([1/v for v in filtered_lb_y])

# Perform linear regression
slope, intercept, r_value, p_value, std_err = linregress(reciprocal_S, reciprocal_V)

# Calculate Km and Vmax from the Lineweaver-Burk plot
Vmax_lb = 1 / intercept
Km_lb = slope * Vmax_lb
R2_lb = r_value**2

print(f"\n--- Lineweaver-Burk Plot Parameters ---")
print(f"Vmax (from LB plot): {Vmax_lb:.4f} U/mg")
print(f"Km (from LB plot): {Km_lb:.4f} mM")
print(f"R-squared (from LB plot): {R2_lb:.4f}")
print(f"Equation of the line: y = {slope:.4f}x + {intercept:.4f}")
print(f"-----------------------------------------")

# Generate points for the fitted line, extending to cross the x-axis
S_fit_lb = np.linspace(min(reciprocal_S.min(), -1/Km_lb * 1.2), max(reciprocal_S), 100)
V_fit_lb = slope * S_fit_lb + intercept

# Create the Lineweaver-Burk plot
plt.figure(figsize=(10, 6))
plt.scatter(reciprocal_S, reciprocal_V, color='blue', label='Experimental Data (Reciprocal)')
plt.plot(S_fit_lb, V_fit_lb, color='green', linestyle='--', label=f'Linear Fit: y = {slope:.4f}x + {intercept:.4f} (R^2 = {R2_lb:.4f})')

# Set Y-axis minimum to zero
plt.ylim(ymin=0)

plt.xlabel('1/[Malate] (1/mM)')
plt.ylabel('1/Activity (1/(U/mg))')
plt.title('Lineweaver-Burk Plot')
plt.grid(True)
plt.legend()
plt.show()
```

# The Eadie-Hofstee plot allows a linear fit without overemphasizing low substrate velocities.

The Eadie-Hofstee plot has v on the Y-axis and v/[S] on the X-axis.

```python
import numpy as np
from scipy.stats import linregress
import matplotlib.pyplot as plt

# Prepare data for Eadie-Hofstee plot
# x_coords are malate concentrations (S)
# y_coords are activities (v)

# Filter out data points where activity or concentration is 0 to avoid division by zero
filtered_eh_x_S = [x for i, x in enumerate(x_coords) if y_coords[i] != 0.0 and x != 0.0]
filtered_eh_y_v = [y for i, y in enumerate(y_coords) if y != 0.0 and x_coords[i] != 0.0]

# Calculate v/[S] for the X-axis
v_over_S = np.array([v / s for v, s in zip(filtered_eh_y_v, filtered_eh_x_S)])

# Perform linear regression on v vs v/[S]
# Eadie-Hofstee equation: v = Vmax - Km * (v/[S])
# Y = Vmax - Km * X
# So, slope (m) = -Km, and Y-intercept (c) = Vmax
slope_eh, intercept_eh, r_value_eh, p_value_eh, std_err_eh = linregress(v_over_S, filtered_eh_y_v)

# Extract parameters
Vmax_eh = intercept_eh
Km_eh = -slope_eh
R2_eh = r_value_eh**2

print(f"\n--- Eadie-Hofstee Plot Parameters ---")
print(f"Vmax (from EH plot): {Vmax_eh:.4f} U/mg")
print(f"Km (from EH plot): {Km_eh:.4f} mM")
print(f"R-squared (from EH plot): {R2_eh:.4f}")
print(f"Equation of the line: v = {intercept_eh:.4f} + ({slope_eh:.4f}) * (v/[S])")
print(f"-----------------------------------------")

# Calculate the x-intercept (where v=0, so Vmax - Km * (v/[S]) = 0 => v/[S] = Vmax / Km)
x_intercept_eh = Vmax_eh / Km_eh if Km_eh != 0 else np.max(v_over_S) * 2

# Generate points for the fitted line, extending to cross both axes
# Start from 0 to ensure crossing the Y-axis
# Extend to cover the maximum v_over_S and the x-intercept (if positive)
max_x_val_for_fit = np.max(v_over_S)
if x_intercept_eh > max_x_val_for_fit:
    max_x_val_for_fit = x_intercept_eh
X_fit_eh = np.linspace(0, max_x_val_for_fit * 1.1, 100) # Extend a bit beyond the farthest point
Y_fit_eh = slope_eh * X_fit_eh + intercept_eh

# Create the Eadie-Hofstee plot
plt.figure(figsize=(10, 6))
plt.scatter(v_over_S, filtered_eh_y_v, color='darkorange', label='Experimental Data (Eadie-Hofstee)')
plt.plot(X_fit_eh, Y_fit_eh, color='blue', linestyle='--', label=f'Linear Fit: v = {intercept_eh:.4f} + ({slope_eh:.4f}) * (v/[S]) (R^2 = {R2_eh:.4f})')

# Set Y- and X-axis minima to zero
plt.ylim(ymin=0)
plt.xlim(xmin=0)

plt.xlabel('v/[S] ( (U/mg) / mM )')
plt.ylabel('v (U/mg)')
plt.title('Eadie-Hofstee Plot')
plt.grid(True)
plt.legend()
plt.show()
```
